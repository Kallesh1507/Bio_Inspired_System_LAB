```python

import random
import math

n = int(input("Enter number of intersections: "))
nests = int(input("Enter number of nests: "))
max_iter = int(input("Enter maximum iterations: "))
pa = float(input("Enter discovery probability (0 to 1): "))

arrival = []
queue = []

for i in range(n):
    print("\nIntersection", i + 1)
    arrival.append(float(input("Enter vehicle arrival rate: ")))
    queue.append(int(input("Enter current queue length: ")))

def fitness(plan):
    total = 0
    for i in range(n):
        waiting = arrival[i] * (plan[i] / 2) + queue[i] / (plan[i] + 1)
        total += waiting + queue[i]
    return total

population = []

for i in range(nests):
    population.append([random.randint(20, 90) for j in range(n)])

scores = [fitness(x) for x in population]
best_index = scores.index(min(scores))
best = population[best_index][:]

for t in range(max_iter):
    for i in range(nests):
        new = population[i][:]
        for j in range(n):
            step = random.gauss(0, 1) * 5
            new[j] = max(10, min(120, int(new[j] + step)))

        new_score = fitness(new)

        if new_score < scores[i]:
            population[i] = new
            scores[i] = new_score

    for i in range(nests):
        if random.random() < pa:
            population[i] = [random.randint(20, 90) for j in range(n)]
            scores[i] = fitness(population[i])

    best_index = scores.index(min(scores))

    if scores[best_index] < fitness(best):
        best = population[best_index][:]

print("\nOPTIMIZED TRAFFIC SIGNAL TIMINGS")
for i in range(n):
    print("Intersection", i + 1, "Green Time:", best[i], "seconds")

print("Minimum Fitness Value:", round(fitness(best), 2))
```
